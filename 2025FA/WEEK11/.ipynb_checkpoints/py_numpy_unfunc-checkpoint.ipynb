```python
sample = [True,False]
arr_b = random.choice(sample,size=5)
arr_b
```

```python
arr1
```

```python
# Example 1 - Sales Trend Analysis:
# Analysis: Calculate the average monthly sales and the percentage chnage
# in sales from the begining to the end of the year.
sales = random.choice(np.arange(100,500), size = 12)
sales
```

```python
average_sales = np.mean(sales)
average_sales
```

```python
percentage_change = ((sales[-1] -sales[0]/sales[0]) * 100)
percentage_change
```
