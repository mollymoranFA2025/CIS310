```python
# Iterating Arrays
arr = np.arange(10)
arr
```

```python
for n in arr:
    print(n)
```

```python
arr2 = arr.reshape(2,5)
arr2
```

```python
for n in arr2:
    print(n)
```

```python
for n in arr2:
    for num in n:
        print(num)
```

```python
for n in np.arange(12).reshape(2,2,3):
    for n1 in n:
        for n2 in n1:
            print(n2)
```

```python
# nditer iterates on each scalar element in the np array
arr = np.arange(24).reshape(2,2,3,2)

for n in np.nditer(arr):
    print(n)
```

```python
# Enumerate Iteration Using ndemerate(), return element and index of the element
for n in np.ndenumerate(arr):
    print(n)
```

```python
arr = np.arange(24).reshape(3,8)
arr
```

```python
for n in np.ndenumerate(arr):
    print(n)
```
