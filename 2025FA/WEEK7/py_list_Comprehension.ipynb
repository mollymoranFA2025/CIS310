### Python list Comprehension

- Python list comprehensions make it easy to create lists, while preforming sophisticated filtering, mapping, and condtional logic on their numbers.
**List Comprehension Format:**
<code> new_list = [expression **for member in iterable**] <code/>
    
**Using Condtional logic**
    
<code> new_list = [expression **for member in iterable**(if conditonal)] <code/>
    
<code> new_list = [expression (if conditional) **for member in iterable**] <code/>

```python
# create a list from a string
lst = []
for letter in 'hello':
    lst.append(letter)
    
lst
```

```python
# [expression for item in list]
lst = [letter for letter in 'hello']
lst
```

```python
squares = []
for n in range(1,11):
    squares.append(n**2)
squares
```

```python
squares = [n**2 for n in range(1,11)]
squares
```

```python
#convert celcuis to fahrenheight
c_lst = [4,10,20,25.5,15.8,0,14.9,-1,-8]

f_lst = [((9/5)*t+32)for t in c_lst]
f_lst
```

```python
# calculate the price after tax for a list of transactions:
txns = [1.09,23.56,57.84,4.56,6.78]
TAX_RATE = .08

final_price = [round(t * (1 + TAX_RATE),2)for t in txns]
final_price
```

```python
# filter the data using condtional logic

# new_list = [expression for member in iterable(if conditonal)] 
squares = [n**2 for n in range(1,11) if n %2 !=0]
squares


```

```python
# find vowels from a sentence
s = "CIS310 communicate students with anyalitical thinking and practical technical skills"

lst = [letter for letter in s if letter.lower() in 'aeiou']
lst
```

```python
# using logical condtion to filter out or map data
#new_list = [expression (if conditional) for member in iterable] 

prices = [1.25,-9.45,10.22,3.78,-5.92,1.16,-8,9,-0.5]
new_prices = [v if v > 0 else 0 for v in prices]
new_prices
```

```python
# using even and odd words
lst = ['even' if n % 2 == 0 else 'odd' for n in range(1,11)]
lst
```

```python
# nested list comprehension
lst = [n**2 for n in range(1,21) if n % 2 !=0]
lst
```

```python
lst = [n **2 for n in [n for n in range(1,21) if n % 2 !=0]]
lst
```

```python
# using Set and dictionary Comprehensions
'''
While the list compnrehension in Python is a common tool, 
you can also create set and dictionary comperhensions.
'''

# set 
s = "Data preperation consumes an amount of time (up to 80%) spent on analytics"
unique_vowels = {l for l in s if l in 'aeiou'}
unique_vowels
```

```python
# dictionary comprehension
# are similar ot list comprehensions, with the addtional requiremnt of defning a key: value pair
d = {n:n**2 for n in range(1,21) if n % 2 !=0}
d
```

```python
# use of list comprehension in dictionary comprehension
cities = ['Austin', 'Tacoma', 'Topeka', 'Scramento','Charlotte']
d = {city:[t for t in range (50,60)] for city in cities}
d
```

```python
from random import randint
d = {city:randint(40,100) for city in cities}
d
```

```python
# elif
lst = [1, 0,2,0,1,1,2,1,0,0]
answers =['yes'if v ==1 else 'no' if v == 2 else 'idle' for v in lst]
answers
```
