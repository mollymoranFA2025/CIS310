```python
# break keyword
n = 1
while n <10:
    print(' n = ', n)
    n+=1
    if n == 5:
        break
else:
    print('n after while loop stops: ', n)
```

```python
#continue key word
n = 1
while n < 10:
    n+=1
    if n ==5:
        continue
    print('n = n', n)
```

```python
#continue key word by using pass
n = 1
while n< 10:
    n+=1
    if n ==5:
        pass
    else:
        print('n = ', n)
```
