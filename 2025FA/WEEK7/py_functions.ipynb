```python
#min_two_odds(3,5) -->
#min_two_odds(num1,num2):
def min_two_odds(num1,num2):
    if (num1 % 2 != 0) and (num2 % 2 !=0):
        return min(num1, num2)
    else:
        return max(num1,num2)
```

```python
min_two_odds(3,5)
```

```python
min_two_odds(4,9)
```

```python
# is _prime(7) ---> 7 is a prime
# is_prime(16) ---> 16 is not a prime
# find a prime number
def is_prime(num):
    for n in range(2,num):
        if num % n == 0:
            return str(num) + ' is not a prime!'
    else:
        return str(num) + ' is a prime!'
        
```

```python
 is_prime(7)
```

```python
is_prime(16)
```

```python
is_prime(9)
```
