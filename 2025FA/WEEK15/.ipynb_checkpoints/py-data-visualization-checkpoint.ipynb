```python
# lime and color style
# using format string such as Colors, markers, and line styles
dataset = randn(50).cumsum()
dataset
```

```python
plt.plot(dataset, color='k')
```

```python
plt.hist(dataset, color='k', bins=15, alpha=0.25)
```

```python
plt.scatter(np.arange(50), dataset) #scatter(x,y)
```

```python
plt.plot(dataset, 'ko--')
```

```python
plt.plot(dataset, color='blue', linestyle='solid',marker='o') # dashed solid for linestyle
```

```python
# Steps
plt.plot(dataset, drawstyle='steps-post')
```

```python
plt.plot(dataset, 'k--')
plt.plot(dataset,drawstyle='steps-post', color='pink')
```

```python
# using labels 
plt.plot(dataset, 'k--', label='Dahsed-Line')
plt.plot(dataset,drawstyle='steps-post', label='Steps-Post')
plt.legend(loc='best')
# locs = [upper, left, lower, center, right, best]
```

```python
# Ticks, Lables, and Legneds
fig = plt.figure()
ax = fig.add_subplot(111)
ax.plot(randn(1000).cumsum())
```

```python
ax.set_xticks(list(range(0,1001,250)))

# labels
ax.set_xticklabels(['one','two','three','four','five'], rotation=30, fontsize='small')

ax.set_title('My first Matplotlip Plot with Ticks')

ax.set_xlabel('Stage')
ax.set_ylabel('Values')

fig
```

```python
# Use a dictionary for title, xlabel, ylabel
props = {
    
    'title': 'My first Matplatlib Plot With ticks',
    'xlabel': 'Stages',
    'ylabel': 'Values'
}
ax.set(**props)
fig
```

```python
# saving a figure in an image file
fig.savefig('myFigure.png')
```
