### Formatting Output with: %, format(), and f-string methods

```python
# escape sequence \
print("Hello\tWorld!") # Inserts a tab between "hello" and "World!"
print("This is a baclslash: \\") # prints a backslash
print("She said, \"Hello'")  # prints: She said "Hello!"
print("Hello\nWorld") #\n is processed to introduce a new line
```

```python
# 1 - Formatting output usin the percent specifier sign %

print("The quick brown %s jumps over the lazy dog" %'cat')
```

```python
print("The quick brown %s jumps over the lazy %s" %('fox', 'dog'))
```

```python
name1, name2, = 'fox', 'dog'
print("The quick brown %s jumps over the lazy %s" %('\tfox','\t#dog'))
```

```python
# Formatting numbers 
print('My score s %s in CIS320' %98.80)
```
