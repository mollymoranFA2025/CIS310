### Python - Pandas DataFrame, Mapping, Applying and Grouping

```python
# import
import numpy as np
from numpy.random import *
import pandas as pd
from pandas import *
```

```python
# mapping  - data points

df = DataFrame({'city': ['Alma', 'Brian Head', 'Fox Park', 'Lake Mary'],
               'altitude': [10355, 9800, 9062, 8966]})
df
```

```python
st = {'Alma': 'Colorado', 'Brain Head':'Utah', 'Fox Park':'Wyoming','Lake Mary':'California'}
st
```

```python
df['state'] = df['city'].map(st)
df
```

```python
#apply
dfc = df.copy()
df['altitude'] = df['altitude'].apply(lambda x: x/ 3.2808)
df
```

```python
df['altitude'] = df['altitude'].apply(lambda x: np.nan if x < 2808 else x)
df
```

```python
df = DataFrame(randn(4,3), columns=list('bde'), index=['UT', 'OH', 'TX', 'OR'])
df
```

```python
df.apply(lambda x:x.max() - x.min())
```

```python
df.apply(lambda x:x.max() - x.min(), axis=1)
```

```python
df.apply(value_counts)
```

```python
# groupby is the most frequently used function in data analysis.
# It is used for grouping the data points (i.e. rows) based on the distinct values

cars = ['Toyota','Ford','GMC','Dodge']
df = DataFrame({'brand': np.random.choice(cars, size=20),
               'price': np.random.randint(40000,85000, size=20)
                
               })
df
```

```python
df['brand'].value_counts()
```

```python
df.groupby('brand').count()
```

```python
df.groupby('brand').sum()
```

```python
rs = df.groupby('brand').mean()
rs
```

```python
df
```

```python
rs = df.groupby('brand')['price'].mean()
rs
```

```python
rs = df.groupby('brand')[['price']].mean() # double brackets is a data frame
rs
```

```python
df
```

```python
df.sort_values(by='brand', inplace=True)
```

```python
df
```

```python
df['cum_sum'] = df.groupby('brand').cumsum()
df
```

```python
# groupby using two columns
sample1 = list('abcd')
sample2 = ['one','two','three','four']

df = DataFrame({
    'k1': np.random.choice(sample1,size=20),
    'k2':np.random.choice(sample2,size=20),
    'dataset1': np.random.randn(20),
    'dataset2': np.random.randn(20),
    'dataset3': np.random.randn(20),
    
})

df
```

```python
df.groupby('k1').count()
```

```python
df.groupby('k1').sum()
```

```python
df.groupby(['k1', 'k2']).sum()
```

```python
df.groupby(['k1', 'k2']).count()
```

```python
df.groupby(['k1', 'k2']).mean

()
```

```python
df.groupby(['k1', 'k2'])[['dataset1']].sum()
```
