```python
#Sample DataFrame with NaN values 
df = DataFrame({'A': np.random.choice(sample,size=10),
                'B': np.random.choice(sample,size=10)
    
    
})
df
```

```python
df.fillna(0)
```

```python
df
```

```python
# Fill NaN values with the previous non-null value
df.fillna(method='ffill') 
```

```python
df['A'].fillna(value=df['A'].mean())
```

```python
# Replace NaNs with the mean of values in the same column
df['A'].fillna(value=df['A'].mean())
```

```python
df
```

```python
df.fillna({'A':df['A'].mean(), 'B': df['B'].median()})
```

```python
df.fillna({'A':df['A'].mean(), 'B': df['B'].median()}, inplace=True)
```

```python
df
```
