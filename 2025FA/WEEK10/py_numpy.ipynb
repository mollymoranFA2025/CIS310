**What is NumPy, and why?**

What:
- NumPy is a Python library.
- NumPy is used for working with arrays.
- NumPy i short for "Numerical Python".

**Why use NumPy**
- Arrays are very frequently used in data science, where speed and resources are very important.
- In Python we have lists that serve the purpose of arrays, but they are slow to process.
- NumPy aims to provide an array of objects that is up to 50x faster than tradtional Python lists.
- The array object in NumPy is called 'ndarray'.
- NumPy is written partailly in Python, but most of the parts that require fast computation are writtenin C or C++
- This Behavior is called locallity of refrence in computer science.
