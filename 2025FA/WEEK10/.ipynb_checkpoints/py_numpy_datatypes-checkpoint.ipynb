```python
# converting Data Type on Exsisitng Arrays
# astype()
arr = np.array([1,2,3,4,5], dtype='S')
arr
```

```python
arr = np.array([1,2,3,4,5], dtype='i4')
arr
```

```python
arr.dtype
```

```python
arr = np.array([1,2,3,4,5], dtype='i8')
arr
```

```python
arr = np.array(['1', '2', '3', '4'], dtype='i')
arr
```

```python
# convert an array after it is created
scores = [random.random() * 100 for _ in range(10)]
scores
```

```python
arr = np.array(scores)
arr
```

```python
arr.dtype
```

```python
arr_int = arr.astype('i8')
arr_int
```

```python
arr_b = arr.astype(bool)
arr_b
```

```python
# NumPy Array Copy as View
# the copy is a new array, and the view is just a view of the original array
# view(), copy(), assignment = 
arr = np.array(range(1,11))
arr
```

```python
arr_assign = arr
arr_assign
```

```python
arr[:2] = [55,55]
arr
```

```python
arr_assign
```

```python
arr_v = arr.view()
arr
```

```python
arr_v
```

```python
arr[-1] = 100
arr
```

```python
arr_v
```

```python
arr_c = arr.copy()
arr_c
```

```python
arr[-2:] = [0,0]
arr
```

```python
arr_c
```

```python
arr_v
```

```python
arr_assign
```

```python
arr.base
```

```python
print(arr.base)
```

```python
print(arr_assign.base)
```

```python
print(arr_v.base)
```

```python
print(arr_c.base)
```

```python
'''
np.vstack: To stack arrays along vertical axis.
np.hstack: To stack arrays along horizontal axis.
'''

a = np.array([[1,2],[3,4]])
b = np.array([[5,6], [7,8]])
```

```python
a.shape
```

```python
b.shape
```

```python
arr_vs = np.vstack((a,b))
arr_vs
```

```python
arr_vs.shape
```

```python
arr_hs = np.hstack((a,b))
arr_hs
```

```python
arr_hs.shape
```

```python
# Reading data from files
f = open("data.txt", "w")
f.write("1, 2, 3\n")
f.write("4, 5, 6")
f.close()
```

```python
f = open('data.txt', 'r')
print(f.read())
```

```python
# Using np.genfromtxt
arr2 = np.genfromtxt('data.txt', dtype=None, delimiter=',')
arr2
```

```python
arr2.shape
```

```python
# Using np.loadtxt
arr2 = np.loadtxt('data.txt', dtype='<U6', delimiter=',')
arr2
```
