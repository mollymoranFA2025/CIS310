### Python - lambda function, higher-order functions: map(), filter(), and reduce()

- A lambda function is a small anonymous function.
- An anyonymous function is a function that is defined without a name.
- They are genrally used when a functionis needed temporarily for a short period of time.
- A lambda function can take any number of arguments, but can only have one expression.

<code> lambda arguments: expression <code/>

```python
def double_values(n):
    return n * 2
```

```python
double_values(4)
```

```python
double_values = lambda num: num * 2
```

```python
double_values(8)
```

```python
# find larger number using a lambda funciton
larger = lambda x,y: x if x> y else y
```

```python
larger(10,15)
```

# Higher-order functions, such as map(), filter(), and reduce()
- A fucntion is called a Higher Order function if it contains other functions as a parameter or returns a function as an output
- A function is an instance of the Object type
- you can store the function in a varible 
- you can pass the function as a parameter to another function

```python
# Python's map() is a build in function
# allows you to process and trnsform all the items in an iterable, without using an explicit for loop
# format: map(function, iterable[,iterable1, iterable2,...., iterableN])
```

```python
# Example:
# using list comprehension
def square(n):
    return n ** 2

squares = [square(n) for n in range(2,21)]
squares
```

```python
list(map(square,range(2,21)))
```

```python
map_lst = map(square,range(2,21))
map_lst
```

```python
list(map_lst)
```

```python
list(map(lambda x: x ** 2,range(2,21)))
```

```python
for item in map(lambda x: x ** 2,range(2,21)):
                print(f"item = {item}")
```

```python
# a list of words' length fora list of words
names = ["Alice", "Bob", "Charlie"]
l = list(map(lambda n: len(n),names))
l
# function -> lambda n: len(n)
# iterable -> names
```

```python
# using map()
import random

txns = [random.random() * 51 for _ in range(10)]
txns
```

```python
TAX_RATE = .08
def gat_price_with_tax(txn):
    return round(txn * (1 + TAX_RATE),2)

final_price = list(map(gat_price_with_tax,txns))
final_price
```

```python
final_price = list(map(lambda txn: round(txn * (1 + TAX_RATE),2),txns))
final_price
```

```python
fahrenheit = [random.random() * 120 for _ in range(10)]
fahrenheit
```

```python
def to_celsius(f):
    return round((5/9)*(f-32),2)

celsius = list(map(to_celsius,fahrenheit))
```

```python
celsius = list(map(lambda n: round((5/9)*(n-32),2),fahrenheit))
celsius
```

```python
# Ditctionary from list comprehnsion
fahrenheit = {'t' + str(i): int(random.random() * 120) for i in range(1,21)}
fahrenheit
```

```python
celsius_dict = {k:to_celsius(v) for (k,v) in fahrenheit.items()}
celsius_dict
```

```python
celsius = list(map(lambda v: round((5/9)*(v-32),2),fahrenheit.values()))
celsius
```

```python
celsius_dict = dict(zip(fahrenheit.keys(),celsius))
celsius_dict
```
