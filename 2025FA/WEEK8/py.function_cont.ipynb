```python
# check code 001 in a list
# [1,3,5,0,0,1,8] -> True
# [1,0,2,5,0,6,1,8] -> true
# [1,5,6,0,6,7,8,0] - false

# define your function
def  check_55(lst):
    code = [0,0,1,'Stop']
    for num in lst:
        if num == code[0]:
            code.pop(0) #remove the number at index 0
            
    return len(code) == 1
```

```python
check_55([1,4,5,0,0,1,8])
```

```python
check_55([1,0,2,5,0,6,1,8])
```

```python
check_55([1,5,6,0,6,7,8,0])
```

```python
# Keyword arguments
# You can also send arguments with the key = valus syntax
# This way the order of the arguments does not matter.
def my_kids(child1, child2, child3): 
    print(f"my first child '{child1}, my second child '{child2}, my third child '{child3}'")

```

```python
# Arbitrary Keywords Arguments, *kwargs
def my_name(**names):
    print(f"nanmes is a dictionary {names}")
    print(f"Last name is {names['lname']}")
    print(f"First name is {names['lname']}")
    print(f"Full name is {names['lname']}, {names['fname']}")
```

```python
my_name(fname = "James", lname = "Smith")
```

```python
my_name(fname = "James")
```

```python
def my_name(**names):
    print(f"names is a dictionary {names}")
    print(f"Last name is {names.get('lname','')}")
    print(f"First name is {names.get('lname','')}")
    print(f"Full name is {names.get('lname','')}, {names.get('fname','')}")
```

```python
my_name(fname = "James")
```
