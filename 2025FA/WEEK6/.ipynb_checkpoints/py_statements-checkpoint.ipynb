### Python Statements [if, else,elseif]

*** the format of the if statment**

```
if bool_condition1:
    preform action1
elif bool_condition2:
    preform action3
else:
    preform action3
```

````python
if True:
    print("It was correct!")
````

````python
if False:
    print("It was correct!")
````

````python
# if else statment 
value = 5 
if value == 5:
    print("Hi Five!")
else:
    print("Let's try again?")
````

````python
# if else statment 
value = 5 
if value == 6:
    print("Hi Five!")
else:
    print("Let's try again?")
````

````python
test1, test2 = 98, 87
if test1 < test2:
    print("Redo test1")
else:
    print("redo test2")
````

````python
#if-elseif statement
test1 = int(input("Enter Test1 score: "))
test2 = int(input("Enter Test2 score: "))

if (test1 > test2):
    print("test1 is greater than test2")
    print("You may think to redo test2!")
elif (test2 > test1): 
    print("test2 is greater then test1")
    print("You may think to redo test1 insetad!")
else:
    print("test1 and test2 are same!")
````

````python
# and logial operator
s1,s2,s3 = 89,75,92
if (s1 > s2) and (s3 > s2):
    print("Both condtions are evaluated to True!")
````

````python
# and logial operator
s1,s2,s3 = 89,75,92
if (s1 < s2) or (s3 > s2):
    print("At least one of the condtions is evaluated to True!")
````

````python
# nested if Statement
s = int(input("Emter value of s: ")) 

if s >= 40:
    print(f"{s} is above or equal 40")
    if s >= 60:
        print(f"{s} is above or equal 60")
    else:
        print(f"{s} is below 40")
elif s <= 30:
    print(f"{s} is above or equal 30")
else:
    print(f"{s} is below 40")
````

````python
# easy way to use built in fuctions instead of if staments
names = {"Fred": 'Hello Fred', 'Rama': 'Hello Rama', 'Joe': 'Hello Joe', 'Alice': 'Hello Alice'}
len(names)
````

````python
name = 'Rama'
if name in names.keys():
    print(names[name])
else:
    print("I don't know who you are!")
````
