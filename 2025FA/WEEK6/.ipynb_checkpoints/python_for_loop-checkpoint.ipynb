````python
sc = [87,90,75,83,79,92,67,56,89,70,81]
sc
````

````python
for s in sc:
    print(f's={s}')
````

````python
for s in sc[::-1]:
    print(f's={s}')
````

````python
sc.reverse()
````

````python
for s in sc:
    if s %2 == 0:
        print(f's={s}')
````

````python
for s in sc:
    if s %2 == 0:
        print(f's is even: {s}')
    else: 
        print(f's is odd: {s}')
````

````python
# fill two lists with even and off values from the sc list
even=[]
odd=[]

for s in sc:
    if s %2 == 0:
        even.append(s)
    else: 
        odd.append(s)
````

````python
# loop over zipped object
lst1 = ['red', 'blue', 'green']
lst2 = [10, 8, 6]
tuple(zip(lst1,lst2))
````

````python
for item in zip(lst1,lst2):
    print(item)
````

````python
for item in zip(lst1,lst2):
    print(item[0], "--->", item[1])
````

````python
for c,n in zip(lst1,lst2):
    print(c, "--->", n)
````

````python
# break and continue
fruits = {'Apple', 'Banana', 'Grape', 'Orange'}
for f in fruits:
    print(f)
````

````python
# break and continue
fruits = ['Apple', 'Banana', 'Grape', 'Orange']
for f in fruits:
    print(f)
    if f == 'Grape':
        break
````

````python
# continue
fruits = ['Apple', 'Banana', 'Grape', 'Orange']
for f in fruits:
    
    if f == 'Grape':
        continue
    #else:
        print(f)
````

````python
# ennumerate
for idx, v in enumerate(fruits):
    print(idx,v)
````

````python
for idx, v in enumerate(fruits, start=1):
    print(idx,v)
````

````python
# what we should do if we need to loop miltiple colletions? Whithout nested loop
colors = ['red', 'green', 'blue', 'purple']
ratios = [0.2,0.3,0.4,0.1]

for idx, v in enumerate(colors):
    print(int(ratios[idx]*100),'%', v)
````

````python
# using range()
for idx in range(len(colors)):
    print(int(ratios[idx]*100),'%', colors[idx])
````

````python
# using zip()
list(zip(ratios,colors))
````

````python
for r,c in zip(ratios,colors):
    print(int(r*100),'%', c)
````
